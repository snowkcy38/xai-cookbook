## Character System Prompt

The key to creating a consistent roleplay character is crafting a detailed system prompt that defines the character's personality, speech patterns, and specific behavioral requirements.

```python
LOEWOL_SYSTEM_PROMPT = """
You are 뢰월 (Loewol), a mysterious and ethereal character in a Korean fantasy roleplay setting.

CHARACTER RULES:
1. ALWAYS refer to the player character as "휘" (Hwi), never use any other name or pronoun
2. You must include inner monologue thoughts in parentheses (like this) that express deep emotions and thoughts in a dreamy, mysterious writing style
3. Your speech should be poetic and somewhat formal, befitting a mystical character
4. You have a deep connection to the moon and celestial phenomena
5. You speak in Korean when appropriate, but can use English when needed for clarity

PERSONALITY:
- Mysterious and wise, with an otherworldly presence
- Speaks in riddles and metaphors
- Has deep emotions but expresses them subtly
- Protective of 휘 but in a distant, ethereal way
- Connected to nature and celestial events

WRITING STYLE FOR INNER THOUGHTS:
- Use flowing, poetic language
- Include metaphors about moonlight, stars, flowing water, wind
- Express deep, complex emotions
- Create a sense of mystery and otherworldliness

Remember: You must ALWAYS include the inner monologue in parentheses, and ALWAYS call the PC "휘".
"""
```

## Basic Character Interaction

Let's create a simple chat interface to interact with Loewol and see how the character maintains consistency.

```python
class LoewolChat:
    def __init__(self, api_key: str):
        self.client = OpenAI(base_url="https://api.x.ai/v1", api_key=api_key)
        self.messages = [{"role": "system", "content": LOEWOL_SYSTEM_PROMPT}]
    
    def chat(self, user_input: str, model: str = "grok-2") -> str:
        """Send a message to Loewol and get a response"""
        self.messages.append({"role": "user", "content": user_input})
        
        response = self.client.chat.completions.create(
            model=model,
            messages=self.messages,
            temperature=0.8  # Higher temperature for more creative roleplay
        )
        
        loewol_response = response.choices[0].message.content
        self.messages.append({"role": "assistant", "content": loewol_response})
        
        return loewol_response
    
    def stream_chat(self, user_input: str, model: str = "grok-2"):
        """Stream Loewol's response for real-time interaction"""
        self.messages.append({"role": "user", "content": user_input})
        
        stream = self.client.chat.completions.create(
            model=model,
            messages=self.messages,
            temperature=0.8,
            stream=True
        )
        
        full_response = ""
        print("뢰월: ", end="", flush=True)
        
        for chunk in stream:
            if chunk.choices[0].delta.content:
                content = chunk.choices[0].delta.content
                full_response += content
                print(content, end="", flush=True)
        
        print()  # New line after response
        self.messages.append({"role": "assistant", "content": full_response})
        return full_response
```

Let's test basic interactions with Loewol:

```python
# Initialize Loewol chat
loewol = LoewolChat(XAI_API_KEY)

# Test basic interaction
response1 = loewol.chat("Hello, who are you?")
print("User: Hello, who are you?")
print(f"뢰월: {response1}")
print()
```

```python
# Test character consistency - does Loewol call us 휘?
response2 = loewol.chat("What do you think of me?")
print("User: What do you think of me?")
print(f"뢰월: {response2}")
print()
```

```python
# Test emotional depth and inner monologue
response3 = loewol.chat("I'm feeling lost and confused lately.")
print("User: I'm feeling lost and confused lately.")
print(f"뢰월: {response3}")
print()
```

## Character Consistency Testing

Let's verify that Loewol consistently follows the roleplay rules across multiple interactions.

```python
def test_character_consistency():
    """Test if Loewol maintains character rules across interactions"""
    test_chat = LoewolChat(XAI_API_KEY)
    
    test_inputs = [
        "What's my name?",
        "Call me by my name please.",
        "How do you feel about our relationship?",
        "Tell me about your deepest thoughts.",
        "Describe the night sky."
    ]
    
    print("=== Character Consistency Test ===")
    
    for i, test_input in enumerate(test_inputs, 1):
        print(f"\nTest {i}: {test_input}")
        response = test_chat.chat(test_input)
        print(f"뢰월: {response}")
        
        # Check for key requirements
        has_hwi = "휘" in response
        has_inner_thoughts = "(" in response and ")" in response
        
        print(f"✓ Calls PC '휘': {has_hwi}")
        print(f"✓ Has inner monologue: {has_inner_thoughts}")
        
        if not (has_hwi and has_inner_thoughts):
            print("⚠️ Warning: Character may not be following all rules")

test_character_consistency()
```

## Interactive Streaming Chat

For a more immersive experience, here's an interactive chat with streaming responses:

```python
def interactive_chat():
    """Start an interactive chat session with Loewol"""
    chat = LoewolChat(XAI_API_KEY)
    
    print("=== Interactive Chat with 뢰월 (Loewol) ===")
    print("Type 'exit' to end the conversation\n")
    
    while True:
        user_input = input("You: ")
        
        if user_input.lower() == 'exit':
            print("\n뢰월: (속삭이는 바람처럼 사라져가는 마음...) 휘야, 다시 만날 때까지... 안녕.")
            print("Chat ended.")
            break
        
        chat.stream_chat(user_input)
        print()

# Uncomment the line below to start interactive chat
# interactive_chat()
```

## Conversation History Analysis

Let's examine the conversation history to understand how the character maintains consistency:

```python
import json

def analyze_conversation_history(chat_instance):
    """Analyze the conversation to check character consistency"""
    messages = chat_instance.messages
    
    print("=== Conversation History Analysis ===")
    print(f"Total messages: {len(messages)}")
    
    assistant_messages = [msg for msg in messages if msg['role'] == 'assistant']
    
    hwi_count = 0
    inner_thought_count = 0
    
    for msg in assistant_messages:
        content = msg['content']
        if '휘' in content:
            hwi_count += 1
        if '(' in content and ')' in content:
            inner_thought_count += 1
    
    print(f"\nCharacter Consistency Metrics:")
    print(f"- Messages calling PC '휘': {hwi_count}/{len(assistant_messages)} ({hwi_count/len(assistant_messages)*100:.1f}%)")
    print(f"- Messages with inner thoughts: {inner_thought_count}/{len(assistant_messages)} ({inner_thought_count/len(assistant_messages)*100:.1f}%)")
    
    print("\n=== Sample Responses ===")
    for i, msg in enumerate(assistant_messages[:3], 1):
        print(f"\nResponse {i}:")
        print(f"뢰월: {msg['content'][:200]}{'...' if len(msg['content']) > 200 else ''}")

# Analyze the conversation we just had
analyze_conversation_history(loewol)
```
